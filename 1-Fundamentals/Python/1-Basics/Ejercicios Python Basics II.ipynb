## Ejercicio 7
Crea un programa que recoja dos inputs. Tiene que comprobar si su suma es igual, superior o inferior a 10

````python
num1 = float(input("introduce un numero: "))
num2 = float(input("introduce un numero: "))

num3 = num1 + num2

if num3 > 10: 
    print("La suma es mayor que 10")
elif num3 == 10: 
    print("la suma es igual a 10")
else:
    print("La suma es menor de 10")
````

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

1. Pásalo todo a mayúsculas
2. Pásalo todo a minúsculas
3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula
4. Crea una lista dividiéndolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

````python
s = "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."
print(s.upper())
print(s.lower())
print(s.title())
print(s.split())
print(s.replace(",", ";"))
print(s.replace("a", ""))

````
